```python
### Exercises 

prices = [10,25,8,40,15]

# 1 numbers of items 

print(len(prices))

# 2 Appl 10% Discount to every price 

discount = []
for i in prices:
    j = i * 0.1
    discount.append(j) 
    
print(discount)    
    
    
discount.sort(reverse=True)
print (discount)
 
```

```python
## Bar Chart

plt.figure(figsize=(6,3))
plt.bar(df['Names'], df['weight'], color='orange')
plt.title("students distribution by weight", fontsize=16, color='black')
plt.xlabel("Names", size=12)
plt.ylabel("Weights", size=12)
plt.xticks(size=11, rotation=0)
plt.yticks(size=11, rotation=0)
plt.show()
```

```python
## Pie Chart 


summary = pd.DataFrame(sl["Gender"].value_counts().reset_index())
summary.columns = ["Gender","Counts"]
summary


plt.title ('Gender Participation', size=16)
plt.pie(summary["Counts"], labels=summary["Gender"], autopct='%1.1f%%')
plt.show()
```

```python
## Histogram

plt.figure (figsize=(8,3))
plt.hist(sl["weight"], bins=2,color='skyblue', edgecolor='black')
plt.title("Distribution of Student Weight")
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.show()
```
